```python
from google.colab import userdata
import cdsapi

CDS_API_KEY = userdata.get('CDS_API_KEY')

client = cdsapi.Client(
    url="https://cds.climate.copernicus.eu/api",
    key=CDS_API_KEY
)

print("CDS client created successfully!")
```

```python
dataset = "reanalysis-era5-land"

request = {
    "variable": ["2m_temperature"],
    "year": "2024",
    "month": "01",
    "day": ["01"],
    "time": ["12:00"],
    "data_format": "netcdf",
    "download_format": "unarchived",

    # GTA bounding box: North, West, South, East
    "area": [44.40, -80.20, 43.30, -78.30]
}

client.retrieve(
    dataset,
    request,
    "GTA_T2M_test_2024_01_01_1200.nc"
)

print("Download completed!")
```

```python
import os
import calendar

# Folder for monthly ERA5-Land files
output_dir = "/content/ERA5Land_T2M_2024"
os.makedirs(output_dir, exist_ok=True)

dataset = "reanalysis-era5-land"

# All 24 hourly timestamps
times = [f"{h:02d}:00" for h in range(24)]

for month in range(1, 13):

    month_str = f"{month:02d}"

    # Correct number of days for each month
    num_days = calendar.monthrange(2024, month)[1]
    days = [f"{d:02d}" for d in range(1, num_days + 1)]

    output_file = os.path.join(
        output_dir,
        f"ERA5Land_T2M_GTA_2024_{month_str}.nc"
    )

    # Skip if already downloaded
    if os.path.exists(output_file):
        print(f"Month {month_str} already exists — skipping.")
        continue

    request = {
        "variable": ["2m_temperature"],
        "year": "2024",
        "month": month_str,
        "day": days,
        "time": times,

        "data_format": "netcdf",
        "download_format": "unarchived",

        # GTA bounding box
        # [North, West, South, East]
        "area": [
            44.40,
            -80.20,
            43.30,
            -78.30
        ]
    }

    print("----------------------------------------")
    print(f"Downloading month {month_str}/2024 ...")

    try:
        client.retrieve(
            dataset,
            request,
            output_file
        )

        print(f"✓ Month {month_str} completed")
        print(f"Saved: {output_file}")

    except Exception as e:
        print(f"✗ Error downloading month {month_str}")
        print(e)

print("----------------------------------------")
print("2024 download loop finished.")
```

```python
import os
import glob
import xarray as xr

folder = "/content/ERA5Land_T2M_2024"

files = sorted(glob.glob(os.path.join(folder, "*.nc")))

print("Number of files found:", len(files))
print("=" * 70)

for file in files:
    ds = xr.open_dataset(file)

    print("\nFile:", os.path.basename(file))
    print("Dimensions:", dict(ds.sizes))
    print("Variables:", list(ds.data_vars))

    # Detect coordinate names
    print("Coordinates:", list(ds.coords))

    ds.close()

print("\n" + "=" * 70)

if len(files) == 12:
    print("✓ All 12 monthly files are present.")
else:
    print(f"⚠ Expected 12 files, but found {len(files)}.")
```
