```python
import os
import calendar

# Folder for monthly ERA5-Land files
output_dir = "/content/ERA5Land_T2M_2024"
os.makedirs(output_dir, exist_ok=True)

dataset = "reanalysis-era5-land"

# All 24 hourly timestamps
times = [f"{h:02d}:00" for h in range(24)]

for month in range(1, 13):

    month_str = f"{month:02d}"

    # Correct number of days for each month
    num_days = calendar.monthrange(2024, month)[1]
    days = [f"{d:02d}" for d in range(1, num_days + 1)]

    output_file = os.path.join(
        output_dir,
        f"ERA5Land_T2M_GTA_2024_{month_str}.nc"
    )

    # Skip if already downloaded
    if os.path.exists(output_file):
        print(f"Month {month_str} already exists — skipping.")
        continue

    request = {
        "variable": ["2m_temperature"],
        "year": "2024",
        "month": month_str,
        "day": days,
        "time": times,

        "data_format": "netcdf",
        "download_format": "unarchived",

        # GTA bounding box
        # [North, West, South, East]
        "area": [
            44.40,
            -80.20,
            43.30,
            -78.30
        ]
    }

    print("----------------------------------------")
    print(f"Downloading month {month_str}/2024 ...")

    try:
        client.retrieve(
            dataset,
            request,
            output_file
        )

        print(f"✓ Month {month_str} completed")
        print(f"Saved: {output_file}")

    except Exception as e:
        print(f"✗ Error downloading month {month_str}")
        print(e)

print("----------------------------------------")
print("2024 download loop finished.")
```

```python
# ============================================================
# ERA5-Land Hourly 2-m Temperature (T2M)
# Greater Toronto Area (GTA)
# Year: 2025
#
# Same spatial domain and settings as the 2024 dataset
# ============================================================


# ============================================================
# STEP 1 — Install required packages
# ============================================================

!pip -q install cdsapi


# ============================================================
# STEP 2 — Import libraries
# ============================================================

import os
import calendar
import cdsapi

from google.colab import userdata


# ============================================================
# STEP 3 — Read CDS API key from Google Colab Secrets
#
# Secret name must be:
# CDS_API_KEY
# ============================================================

CDS_API_KEY = userdata.get("CDS_API_KEY")

print("✓ CDS API key loaded from Colab Secrets")


# ============================================================
# STEP 4 — Create CDS configuration file
# ============================================================

cdsapirc_path = "/root/.cdsapirc"

with open(cdsapirc_path, "w") as f:
    f.write("url: https://cds.climate.copernicus.eu/api\n")
    f.write(f"key: {CDS_API_KEY}\n")

os.chmod(cdsapirc_path, 0o600)

print("✓ CDS configuration file created")


# ============================================================
# STEP 5 — Create CDS client
# ============================================================

client = cdsapi.Client()

print("✓ CDS API client initialized successfully")


# ============================================================
# STEP 6 — Create output folder
# ============================================================

output_dir = "/content/ERA5Land_T2M_2025"

os.makedirs(
    output_dir,
    exist_ok=True
)

print("\nOutput folder:")
print(output_dir)


# ============================================================
# STEP 7 — ERA5-Land dataset
# ============================================================

dataset = "reanalysis-era5-land"


# ============================================================
# STEP 8 — Define all 24 hourly timestamps
# ============================================================

times = [
    f"{h:02d}:00"
    for h in range(24)
]


# ============================================================
# STEP 9 — GTA bounding box
#
# IMPORTANT:
# This is EXACTLY the same bounding box used for 2024.
#
# Format:
# [North, West, South, East]
# ============================================================

GTA_AREA = [
    44.40,     # North
    -80.20,    # West
    43.30,     # South
    -78.30     # East
]

print("\nGTA bounding box:")
print(GTA_AREA)


# ============================================================
# STEP 10 — Download each month separately
# ============================================================

for month in range(1, 13):

    month_str = f"{month:02d}"

    # --------------------------------------------------------
    # Determine correct number of days for each month
    # --------------------------------------------------------

    num_days = calendar.monthrange(
        2025,
        month
    )[1]

    days = [
        f"{d:02d}"
        for d in range(1, num_days + 1)
    ]


    # --------------------------------------------------------
    # Output filename
    # --------------------------------------------------------

    output_file = os.path.join(
        output_dir,
        f"ERA5Land_T2M_GTA_2025_{month_str}.nc"
    )


    print("\n" + "=" * 65)
    print(f"Month: {month_str}/2025")
    print("=" * 65)


    # --------------------------------------------------------
    # Skip month if already downloaded
    # --------------------------------------------------------

    if os.path.exists(output_file):

        print("✓ File already exists.")
        print("Skipping this month:")
        print(output_file)

        continue


    # --------------------------------------------------------
    # CDS request
    # --------------------------------------------------------

    request = {

        "variable": [
            "2m_temperature"
        ],

        "year": "2025",

        "month": month_str,

        "day": days,

        "time": times,

        "data_format": "netcdf",

        "download_format": "unarchived",

        # GTA bounding box
        # [North, West, South, East]

        "area": GTA_AREA
    }


    # --------------------------------------------------------
    # Download
    # --------------------------------------------------------

    print(f"Downloading T2M for {month_str}/2025 ...")

    try:

        client.retrieve(
            dataset,
            request,
            output_file
        )

        print(f"✓ Month {month_str} completed")
        print(f"Saved:")
        print(output_file)


    except Exception as e:

        print(f"✗ Error downloading month {month_str}")
        print(e)


# ============================================================
# STEP 11 — Check downloaded files
# ============================================================

print("\n")
print("=" * 65)
print("CHECKING 2025 FILES")
print("=" * 65)

import glob

files = sorted(
    glob.glob(
        os.path.join(
            output_dir,
            "*.nc"
        )
    )
)

print("\nNumber of files found:", len(files))

for file in files:
    print("✓", os.path.basename(file))


# ============================================================
# STEP 12 — Final verification
# ============================================================

print("\n" + "=" * 65)

if len(files) == 12:

    print("✓ SUCCESS")
    print("All 12 monthly T2M files for 2025 are present.")

else:

    print("⚠ WARNING")
    print(
        f"Expected 12 monthly files, "
        f"but found {len(files)}."
    )

print("=" * 65)
```

```python
# ============================================================
# ERA5-Land Hourly 10-m Wind Components
# U10 + V10
# Greater Toronto Area (GTA)
# Years: 2024 and 2025
# ============================================================


# ============================================================
# STEP 1 — Install CDS API
# ============================================================

!pip -q install cdsapi


# ============================================================
# STEP 2 — Import libraries
# ============================================================

import os
import calendar
import glob
import cdsapi

from google.colab import userdata


# ============================================================
# STEP 3 — Load CDS API key from Colab Secrets
# ============================================================

CDS_API_KEY = userdata.get("CDS_API_KEY")

print("✓ CDS API key loaded from Colab Secrets")


# ============================================================
# STEP 4 — Create CDS configuration
# ============================================================

with open("/root/.cdsapirc", "w") as f:
    f.write("url: https://cds.climate.copernicus.eu/api\n")
    f.write(f"key: {CDS_API_KEY}\n")

os.chmod("/root/.cdsapirc", 0o600)

print("✓ CDS configuration created")


# ============================================================
# STEP 5 — Initialize CDS client
# ============================================================

client = cdsapi.Client()

print("✓ CDS API client initialized")


# ============================================================
# STEP 6 — ERA5-Land dataset
# ============================================================

dataset = "reanalysis-era5-land"


# ============================================================
# STEP 7 — GTA bounding box
#
# EXACTLY the same area used for T2M and D2M
#
# Format:
# [North, West, South, East]
# ============================================================

GTA_AREA = [
    44.40,     # North
    -80.20,    # West
    43.30,     # South
    -78.30     # East
]


# ============================================================
# STEP 8 — All 24 hourly timestamps
# ============================================================

times = [
    f"{h:02d}:00"
    for h in range(24)
]


# ============================================================
# STEP 9 — Download U10 + V10
# for 2024 and 2025
# ============================================================

for year in [2024, 2025]:

    print("\n")
    print("#" * 70)
    print(f"STARTING U10 + V10 DOWNLOAD FOR {year}")
    print("#" * 70)

    # Separate folder for each year
    output_dir = f"/content/ERA5Land_Wind_U10_V10_{year}"

    os.makedirs(
        output_dir,
        exist_ok=True
    )


    # ========================================================
    # Monthly loop
    # ========================================================

    for month in range(1, 13):

        month_str = f"{month:02d}"


        # ----------------------------------------------------
        # Correct number of days for each month
        # Automatically handles leap year 2024
        # ----------------------------------------------------

        num_days = calendar.monthrange(
            year,
            month
        )[1]

        days = [
            f"{d:02d}"
            for d in range(1, num_days + 1)
        ]


        # ----------------------------------------------------
        # Output filename
        # ----------------------------------------------------

        output_file = os.path.join(
            output_dir,
            f"ERA5Land_U10_V10_GTA_{year}_{month_str}.nc"
        )


        print("\n" + "=" * 65)
        print(f"U10 + V10 — {year}-{month_str}")
        print("=" * 65)


        # ----------------------------------------------------
        # Skip if already downloaded
        # ----------------------------------------------------

        if os.path.exists(output_file):

            print("✓ File already exists — skipping.")
            print(output_file)

            continue


        # ----------------------------------------------------
        # CDS request
        # ----------------------------------------------------

        request = {

            "variable": [

                "10m_u_component_of_wind",

                "10m_v_component_of_wind"

            ],

            "year": str(year),

            "month": month_str,

            "day": days,

            "time": times,

            "data_format": "netcdf",

            "download_format": "unarchived",

            "area": GTA_AREA
        }


        # ----------------------------------------------------
        # Download
        # ----------------------------------------------------

        print(
            f"Downloading U10 + V10 for "
            f"{year}-{month_str} ..."
        )

        try:

            client.retrieve(
                dataset,
                request,
                output_file
            )

            print(
                f"✓ {year}-{month_str} completed"
            )

            print(
                f"Saved: {output_file}"
            )


        except Exception as e:

            print(
                f"✗ Error downloading "
                f"{year}-{month_str}"
            )

            print(e)


# ============================================================
# STEP 10 — Verify downloaded files
# ============================================================

print("\n")
print("#" * 70)
print("FINAL WIND DATA CHECK")
print("#" * 70)


for year in [2024, 2025]:

    folder = f"/content/ERA5Land_Wind_U10_V10_{year}"

    nc_files = sorted(
        glob.glob(
            os.path.join(
                folder,
                "*.nc"
            )
        )
    )

    print("\n" + "=" * 60)

    print(
        f"{year}: "
        f"{len(nc_files)} monthly files found"
    )

    print("=" * 60)


    for file in nc_files:

        print(
            "✓",
            os.path.basename(file)
        )


    if len(nc_files) == 12:

        print(
            f"\n✓ SUCCESS: "
            f"All 12 U10 + V10 files "
            f"for {year} are present."
        )

    else:

        print(
            f"\n⚠ WARNING: Expected 12 files "
            f"for {year}, but found "
            f"{len(nc_files)}."
        )


print("\n" + "#" * 70)
print("U10 + V10 DOWNLOAD PROCESS FINISHED")
print("#" * 70)
```
